## Source Data Loading

Load each workbook sheet into its own dataframe. This section only reads the raw source data; merging and cleaning happen below.

```python
source_path = "data/raw/Data.xlsm"
df1 = pd.read_excel(source_path, sheet_name="DataV2")
df2 = pd.read_excel(source_path, sheet_name="Binned")

print("DataV2:", df1.shape)
print("Binned:", df2.shape)
print("Merge key available:", "Customer" in df1.columns and "Customer" in df2.columns)
```

## Load Data

Read the two source files — the main data and the additional-info file — into separate dataframes before combining them.

Data Merging 

Join the two dataframes on the `Customer` key and compare the four join types (inner, left, right, outer) by their resulting row counts.

```python
# 2. The column to join on (must exist in both files)
key = "Customer"

# 3. Do the four joins
inner = pd.merge(df1, df2, on=key, how="inner")
left  = pd.merge(df1, df2, on=key, how="left")
right = pd.merge(df1, df2, on=key, how="right")
outer = pd.merge(df1, df2, on=key, how="outer")



print("inner:", inner.shape)
print("left: ", left.shape)
print("right:", right.shape)
print("outer:", outer.shape)
```

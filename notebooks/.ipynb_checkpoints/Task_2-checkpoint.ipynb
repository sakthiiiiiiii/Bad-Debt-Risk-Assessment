```python
### Task 3 — Feature Importance and Impact Analysis

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder

# Create integer-encoded target for modeling and OneHotEncoder for multiclass AUC
if 'OUTCOME_ENC' not in data.columns:
    data['OUTCOME_ENC'] = data['OUTCOME'].astype('category').cat.codes
if 'outcome_ohe' not in globals():
    ohe = OneHotEncoder(sparse=False, handle_unknown='ignore')
    ohe.fit(data['OUTCOME'].values.reshape(-1, 1))
    outcome_ohe = pd.DataFrame(ohe.transform(data['OUTCOME'].values.reshape(-1, 1)),
                               columns=ohe.get_feature_names_out(), index=data.index)
    # mapping from integer-encoded classes (0..n-1) to one-hot column names
    categories = data['OUTCOME'].astype('category').cat.categories.tolist()
    col_map = {i: ohe.get_feature_names_out()[i] for i in range(len(categories))}

feature_cols = [c for c in data.columns if c not in ['Customer', 'OUTCOME', 'OUTCOME_ENC'] and data[c].dtype in ['int64', 'float64']]

X = data[feature_cols]
y = data['OUTCOME_ENC']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=42)

rf = RandomForestClassifier(n_estimators=200, random_state=42)
rf.fit(X_train, y_train)

importances = pd.Series(rf.feature_importances_, index=feature_cols).sort_values(ascending=False)

print('Top 10 important features:')
print(importances.head(10))

print('\nLowest 10 important features:')
print(importances.tail(10))

plt.figure(figsize=(12, 6))
sns.barplot(x=importances.head(10).values, y=importances.head(10).index, palette='viridis')
plt.title('Top 10 Feature Importances')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x=importances.tail(10).values, y=importances.tail(10).index, palette='magma')
plt.title('Lowest 10 Feature Importances')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()

print('Interpretation: features with the highest importance are the strongest predictors of the target. Features with the lowest importance have little effect on the model output and may be candidates for removal in a simplified model.')
```

```python
# Robust Task 4: model comparison with multiclass-safe AUC handling
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

# Ensure integer target and OneHotEncoder-based one-hot target for multiclass AUC
from sklearn.preprocessing import OneHotEncoder
if 'OUTCOME_ENC' not in data.columns:
    data['OUTCOME_ENC'] = data['OUTCOME'].astype('category').cat.codes
if 'outcome_ohe' not in globals():
    ohe = OneHotEncoder(sparse=False, handle_unknown='ignore')
    ohe.fit(data['OUTCOME'].values.reshape(-1, 1))
    outcome_ohe = pd.DataFrame(ohe.transform(data['OUTCOME'].values.reshape(-1, 1)),
                               columns=ohe.get_feature_names_out(), index=data.index)
    categories = data['OUTCOME'].astype('category').cat.categories.tolist()
    col_map = {i: ohe.get_feature_names_out()[i] for i in range(len(categories))}

# Feature groups
feature_sets = {
    'Credit Scores': ['RISK_SCORE', 'NO_SCORE', 'SCORE_CR22'],
    'Risk Features': [col for col in data.columns if 'RISK_' in col and col != 'RISK_RELATIVE_CR21'],
    'Selected Features': selected_features if 'selected_features' in globals() else [],
    'SFM Selected': selected_sfm if 'selected_sfm' in globals() else [],
    'Risk Relative Index': ['RISK_RELATIVE_CR21'],
    'Credit + Risk': ['RISK_SCORE', 'NO_SCORE', 'SCORE_CR22', 'RISK_RELATIVE_CR21']
}

# Keep only numeric, present columns
numeric_cols = set(data.select_dtypes(include=[np.number]).columns)
feature_sets = {name: [c for c in cols if c in numeric_cols] for name, cols in feature_sets.items()}
feature_sets = {name: cols for name, cols in feature_sets.items() if cols}

models = {
    'Logistic Regression': LogisticRegression(max_iter=2000, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'HistGradientBoosting': HistGradientBoostingClassifier(random_state=42)
}

results = []
trained_models = {}

for set_name, features in feature_sets.items():
    X = data[features].copy()
    y = data['OUTCOME_ENC']

    # Drop rows missing these features
    valid_idx = X.dropna().index
    X = X.loc[valid_idx]
    y = y.loc[valid_idx]
    # select rows from precomputed OneHotEncoder transform
    y_ohe = outcome_ohe.loc[valid_idx]

    if y.nunique() < 2:
        print(f"Skipping {set_name}: only one class after dropna")
        continue

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=42)

    for model_name, model in models.items():
        clf = model
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)

        # obtain probability matrix (n_samples, n_classes)
        if hasattr(clf, 'predict_proba'):
            y_proba = clf.predict_proba(X_test)
        else:
            df_dec = clf.decision_function(X_test)
            if np.ndim(df_dec) == 1:
                probs_pos = 1/(1+np.exp(-df_dec))
                y_proba = np.vstack([1-probs_pos, probs_pos]).T
            else:
                ex = np.exp(df_dec - np.max(df_dec, axis=1, keepdims=True))
                y_proba = ex / ex.sum(axis=1, keepdims=True)

        # compute AUC safely for binary and multiclass (use one-hot for multiclass)
        try:
            if y_test.nunique() == 2:
                pos_col = 1 if y_proba.shape[1] > 1 else 0
                score_for_auc = y_proba[:, pos_col]
                auc = roc_auc_score(y_test, score_for_auc)
            else:
                # select one-hot rows for X_test and reorder to match classifier classes
                y_test_ohe = y_ohe.loc[X_test.index]
                cols = [col_map[c] for c in clf.classes_]
                y_test_ohe = y_test_ohe[cols]
                auc = roc_auc_score(y_test_ohe, y_proba, multi_class='ovr')
        except Exception as e:
            auc = np.nan
            print('AUC error for', set_name, model_name, '->', e)

        report = classification_report(y_test, y_pred, output_dict=True, zero_division=0)
        acc = report.get('accuracy', np.nan)

        results.append({
            'Feature Set': set_name,
            'Model': model_name,
            'AUC': auc,
            'Accuracy': acc
        })
        trained_models[(set_name, model_name)] = (clf, X_test, y_test, y_proba)

result_df = pd.DataFrame(results)
if not result_df.empty:
    result_df = result_df.sort_values(by=['AUC', 'Accuracy'], ascending=False)

result_df
```

```python
best_row = result_df.iloc[0]
best_key = (best_row['Feature Set'], best_row['Model'])
best_model, X_test, y_test, y_proba = trained_models[best_key]
print('Best model:')
print(best_row)
print('\nConfusion matrix:')
print(confusion_matrix(y_test, best_model.predict(X_test)))

# Ensure SHAP is available
try:
    import shap
except Exception:
    import sys
    !{sys.executable} -m pip install shap
    import shap

# Create an explainer appropriate for the model type
if hasattr(best_model, 'predict_proba') and best_model.__class__.__name__ in ['RandomForestClassifier', 'HistGradientBoostingClassifier']:
    explainer = shap.TreeExplainer(best_model)
else:
    explainer = shap.LinearExplainer(best_model, X_test, feature_dependence='independent')

# compute SHAP values (handle multiclass outputs)
shap_values = explainer.shap_values(X_test)

plt.figure(figsize=(10, 6))
if isinstance(shap_values, list):
    # multiclass: show class 1 (or the last) summary
    try:
        shap.summary_plot(shap_values[1], X_test, show=False)
    except Exception:
        shap.summary_plot(shap_values[-1], X_test, show=False)
else:
    shap.summary_plot(shap_values, X_test, show=False)
plt.tight_layout()
plt.show()
```

```python
# ROC curve and AUC for the selected model (binary or multiclass)
from sklearn.metrics import roc_curve, auc

plt.figure(figsize=(8, 6))
if y_test.nunique() == 2:
    pos = 1 if y_proba.shape[1] > 1 else 0
    scores = y_proba[:, pos]
    fpr, tpr, _ = roc_curve(y_test, scores)
    roc_auc = roc_auc_score(y_test, scores)
    plt.plot(fpr, tpr, label=f'AUC = {roc_auc:.4f}', color='navy')
else:
    # multiclass: plot per-class ROC curves and micro-average
    y_test_ohe = outcome_ohe.loc[X_test.index]
    for i, col in enumerate(y_test_ohe.columns):
        fpr, tpr, _ = roc_curve(y_test_ohe.iloc[:, i], y_proba[:, i])
        plt.plot(fpr, tpr, label=f'{col} (AUC={auc(fpr,tpr):.3f})')
    # micro-average AUC
    try:
        micro_auc = roc_auc_score(y_test_ohe, y_proba, average='micro', multi_class='ovr')
        plt.plot([], [], label=f'micro AUC = {micro_auc:.4f}')
    except Exception:
        pass

plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title(f'ROC Curve — {best_row["Model"]} on {best_row["Feature Set"]}')
plt.legend(loc='lower right')
plt.grid(True)
plt.tight_layout()
plt.show()
```
